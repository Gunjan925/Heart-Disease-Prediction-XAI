## Converting data into numerical format for the ML model

```python
df['ca'] = df['ca'].apply(pd.to_numeric)
df['thal'] = df['thal'].apply(pd.to_numeric)
df.info()
```

## Dividing cols into numerical , binary and encoding

```python
num_cols = ['age','trestbps','chol','thalach','oldpeak','ca']
binary_cols = ['fbs','exang']
categorical_cols = ['cp','restecg','slope','thal']
```

## Making maps for columns

```python
## binary_map (yes:1,no:0)
binary_map = {"Yes":1 , "No":0}
## for gender (1 = male; 0 = female)
gender_map = {"Male":1,"Female":0}
## for chest pain -- Value 1: typical angina -- Value 2: atypical angina -- Value 3: non-anginal pain -- Value 4: asymptomatic
chest_pain_map = {"Typical Angina":1 , "Atypical Angina": 2 , "Non-anginal Pain": 3 , "Asymptomatic": 4}
# for restecg -- Value 0: normal -- Value 1: having ST-T wave abnormality (T wave inversions and/or ST elevation or depression of > 0.05 mV) -- Value 2: showing probable or definite left ventricular hypertrophy by Estes' criteria
restecg_map = {"Normal": 0 , "ST-T Wave Abnormality": 1 , "Left Ventricular Hypertrophy": 2}
## slope -- Value 1: upsloping -- Value 2: flat -- Value 3: downsloping
slope_map = {"Upsloping": 1 , "Flat": 2 , "Downsloping": 3}
## thal 3 = normal; 6 = fixed defect; 7 = reversable defect
thal_map = {"Normal": 3 , "Fixed Defect": 6 , "Reversible Defect": 7}
```

## Dividing the data into train and test

```python
from sklearn.model_selection import train_test_split

X = df.iloc[:,:-1]
y = df.iloc[:,-1]

X_train , X_test , y_train , y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

## Encoding data

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_encoder = OneHotEncoder(drop='first',sparse_output=False,handle_unknown='ignore')
one_hot_encoder.fit(X_train[categorical_cols])

X_train_encoded = one_hot_encoder.transform(X_train[categorical_cols])
X_test_encoded = one_hot_encoder.transform(X_test[categorical_cols])
encoded_cols = one_hot_encoder.get_feature_names_out()
X_train_encoded_df = pd.DataFrame(X_train_encoded,columns=encoded_cols,index=X_train.index)
X_test_encoded_df = pd.DataFrame(X_test_encoded,columns=encoded_cols,index=X_test.index)

X_train = X_train.drop(categorical_cols,axis=1)
X_test = X_test.drop(categorical_cols,axis=1)
X_train = pd.concat([X_train,X_train_encoded_df],axis=1)
X_test = pd.concat([X_test,X_test_encoded_df],axis=1)

X_train.head()
```

## Saving encoder

```python
import joblib
joblib.dump(one_hot_encoder,'../../../models/heart/new_models/new_one_hot_encoder.pkl')
```

## Scaling data

```python
from sklearn.preprocessing import StandardScaler

standard_scaler = StandardScaler()
standard_scaler.fit(X_train[num_cols])

X_train[num_cols] = standard_scaler.transform(X_train[num_cols])
X_test[num_cols] = standard_scaler.transform(X_test[num_cols])

X_train.head()
```

## Saving scaler

```python
import joblib
joblib.dump(standard_scaler,'../../../models/heart/new_models/new_standard_scaler.pkl')
```

## Saving test and train datasets

```python
X_train.to_csv("../../../data/heart/processed/new_dataset_cleveland/X_train.csv",index=False)
X_test.to_csv("../../../data/heart/processed/new_dataset_cleveland/X_test.csv",index=False)
y_train.to_csv("../../../data/heart/processed/new_dataset_cleveland/y_train.csv",index=False)
y_test.to_csv("../../../data/heart/processed/new_dataset_cleveland/y_test.csv",index=False)
```
